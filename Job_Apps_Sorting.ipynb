```python
def before_original(a,b):

  ranknum_A = float(a[2])*0.65 + float(a[3])*.4 + float(a[4])*5 + float(a[5])*5 + float(a[6])*1
  ranknum_B = float(b[2])*0.65 + float(b[3])*.4 + float(b[4])*5 + float(b[5])*5 + float(b[6])*1


#level

  if a[7] == "proficient":
    ranknum_B += 15000
  elif a[7] == "advanced":
    ranknum_B += 100
  elif a[7] == "developing":
    ranknum_B += 0

  if b[7] == "proficient":
    ranknum_B += 15000
  elif b[7] == "advanced":
    ranknum_B += 100
  elif b[7] == "developing":
    ranknum_B += 0


#language

  if a[8] == "Python":
    ranknum_A += 150
  elif a[8] == "JavaScript":
    ranknum_A += 100
  elif a[8] == "Java":
    ranknum_A += 100

  if b[8] == "Python":
    ranknum_B += 15000
  elif b[8] == "JavaScript":
    ranknum_B += 100
  elif b[8] == "Java":
    ranknum_B += 100


#specialty

  if a[9] == "data":
    ranknum_A += 150
  elif a[9] == "web":
    ranknum_A += 100
  elif a[9] == "backend":
    ranknum_A += 100

  if b[9] == "data":
    ranknum_B += 150
  elif b[9] == "web":
    ranknum_B += 50
  elif b[9] == "backend":
    ranknum_B += 200


#location

  if a[10] == "onsite":
    ranknum_A += 300
  elif a[10] == "hybrid":
    ranknum_A += 250
  elif a[10] == "remote":
    ranknum_A += 50

  if b[10] == "onsite":
    ranknum_B += 300
  elif b[10] == "hybrid":
    ranknum_B += 250
  elif b[10] == "remote":
    ranknum_B += 50


  #print(f"A: {ranknum_A}")
  #print(f"B: {ranknum_B}")

  if ranknum_B > ranknum_A:
    return True
  elif ranknum_B < ranknum_A:
    return False
  else:
    int(a[0]) > int(b[0])


  print(ranknum_A)
  print(ranknum_B)

before_original(data[2],data[3])
```

```python
from typing import dataclass_transform
data = data[1:]
def partition(data, low, high):
  pivot = data[high]
  i = low - 1

  for j in range(low, high):
     if before_original(data[j],pivot):
       i += 1
       data[i], data[j] = data[j], data[i]

  data[i+1], data[high] = data[high], data[i+1]
  return i+1

def quicksort(data, low=0, high=None):
  if high is None:
    high = len(data) - 1

  if low < high:
    pivot_index = partition(data, low, high)
    quicksort(data, low, pivot_index-1)
    quicksort(data, pivot_index+1, high)

quicksort(data)
print(data[:-1])
```

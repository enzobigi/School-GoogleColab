```python
import csv
import pandas as pd
```

```python
data = []
with open("data.csv") as file:
  csv_reader = csv.reader(file)
  for row in csv_reader:
    data.append(row)
```

```python
print(data[0])

print(data[1])
print(data[2])
print(data[3])

print(data[1][3])
```

```python
def before_original(a,b):
  ranknum_A = float(data[a][2])*0.65 + float(data[a][3])*0.4 + float(data[a][4])*.3 + float(data[a][5])*.4 + float(data[a][5])*0.1
  ranknum_B = float(data[b][2])*0.65 + float(data[b][3])*0.4 + float(data[b][4])*.3 + float(data[b][5])*.4 + float(data[b][5])*0.1

  print(f"A: {ranknum_A}")
  print(f"B: {ranknum_B}")

  if a > b:
    return True
  elif a <= b:
    return False


  print(ranknum_A)
  print(ranknum_B)

before_original(1,5000)
```

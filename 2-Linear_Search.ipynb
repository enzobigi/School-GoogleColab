```python
def linear_search(numbers, target):
  targetsfound = []
  diff = float('inf')
  for i in range(len(numbers)):
    if numbers[i] == target:
      targetsfound.append(i)
    else:
      currentdistance = abs(target - numbers[i])
      if currentdistance < diff:
        diff = currentdistance
        closestnumber = numbers[i]

  if len(targetsfound) > 0:
    return f"Your target, {target}, was found {len(targetsfound)} times at indexes {targetsfound}"
  return f"Your number, {target}, was not found, but, the closest number to it was {closestnumber}."

numbers = [2,27,27,33,55,57,67,921]
target = 24


linear_search(numbers, target)
```

```python
#PLURALITY


votingresults = [
    "DABC", "DACB", "CADB", "CDAB", "BDAC", "ADBC", "DCBA", "DCAB", "CABD", "DCAB",
    "DBCA", "DCAB", "CDAB", "BDAC", "DBCA", "CDAB", "BCAD", "CDBA", "ABCD", "CADB",
    "DABC", "ADBC", "ACBD", "ABDC"
]

counts = []
for vote in votingresults:
  if vote in counts:
    counts[vote] += 1
  else:
    counts[vote] = 1


print(counts)


```

```python
#RUN-OFF

votingresults = [
    "DABC", "DACB", "CADB", "CDAB", "BDAC", "ADBC", "DCBA", "DCAB", "CABD", "DCAB",
    "DBCA", "DCAB", "CDAB", "BDAC", "DBCA", "CDAB", "BCAD", "CDBA", "ABCD", "CADB",
    "DABC", "ADBC", "ACBD", "ABDC"
]

a = 0
b = 0
c = 0
d = 0

for i in range(len(votingresults)):
  if votingresults[i][0] == "A":
    a += 1
  elif votingresults[i][0] == "B":
    b += 1
  elif votingresults[i][0] == "C":
    c += 1
  elif votingresults[i][0] == "D":
    d += 1

#sorting

totalvotes = a + b + c + d
votes = {"A" : int(a),
         "B" : int(b),
         "C" : int(c),
         "D" : int(d)
         }
minimumvotes = votes[1]

for i in range(len(votes)):
  if votes[1] < minimumvotes:
    votes[1] = minimumvotes



print(votes)
```

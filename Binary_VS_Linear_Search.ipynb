```python
list1 = list(range(1,10000))
list1.sort()

number = 0

low = 0
high = len(list1) - 1

def linear_search(list1, number):
  linearoperations = 0
  for i in list1:
    linearoperations += 1
    if list1[i] == number:
      print (f"(Linear Search - {number}) You found the number!")
      print (f"It took you {linearoperations} tries.")
      break
    elif i == len(list1):
      print ("Number was not found")


def binary_search(list1, number, high, low):
  binaryoperations = 0
  while True:
    mid = (high + low)//2
    if list1[mid] == number:
      print (f"(Binary Search - {number}) You found the number!")
      print (f"It took you {binaryoperations} tries.")
      break
    elif list1[mid] > number:
      high = mid - 1
      binaryoperations += 1
    elif list1[mid] < number:
      low = mid + 1
      binaryoperations += 1
    elif list1[low] == list1[high]:
      print ("Number was not found")
      break

binary_search(list1, 2345, high, low)
linear_search(list1, 2345,)
print("")
binary_search(list1, 2335, high, low)
linear_search(list1, 2335,)
```

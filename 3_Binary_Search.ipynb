```python
def binarysearch(numbers, target):
  low = 0
  high = len(numbers) - 1

  foundcount = 0

  if target > max(numbers):
    print (f"Your target was not found. The closest value in the list was {max(numbers)}")

  if target < min(numbers):
    print (f"Your target was not found. The closest value in the list was {min(numbers)}")

  while low <= high:
    mid = (low + high) // 2

    if numbers[mid] == target:
      foundcount = foundcount + 1
      print (f"Your target, {target}, was found at index {mid}!")
      print (f"Total found so far: {foundcount}")
      numbers.pop(mid)
      print (numbers)


    #IF TARGET WAS NOT FOUND, SHRINK THE LIST BY SPLITTING RANGE IN HALF

    elif numbers[mid] > target:
      high = mid - 1

    else:
      low = mid + 1

#IF TARGET WAS NOT FOUND (I put this outside of the while loop because low is already greater than high. either high or low is now placed outside)

  if high < 0:
    print (f"Your target was not found, but the closest valid result to it was {numbers[high]}")

  elif low >= len(numbers):
    print (f"Your target was not found, but the closest valid result to it was {numbers[low]}")

  elif foundcount == 0:
    #FINDING WHICH OF THE TWO OPTIONS LEFT ARE CLOSER
    distanceofhightotarget = abs(target - numbers[high])
    distanceoflowtotarget = abs(numbers[low] - target)
    if distanceofhightotarget < distanceoflowtotarget:
      print (f"Your target was not found. The closest value in the list was {numbers[high]}")
    elif distanceoflowtotarget < distanceofhightotarget:
      print (f"Your target was not found. The closest value in the list was {numbers[low]}")
    else:
      print (f"Your target was not found. The closest values in the list are {numbers[high]} and {numbers[low]}")


numbers = [1,2,3,4,5,6]
target = 3

binarysearch(numbers, target)
```

```python
votes = [["steak","chicken","fish"],["steak","fish","chicken"],["chicken","fish","steak"],["fish","chicken","steak"],["fish","steak","chicken"],["steak","chicken","fish"],["steak","chicken","fish"],["steak","fish","chicken"],["fish","chicken","steak"],["chicken","fish","steak"],["chicken","fish","steak"],["steak","fish","chicken"],["fish","chicken","steak"],["fish","steak","chicken"],["chicken","steak","fish"],["steak","fish","chicken"],["fish","chicken","steak"],["steak","fish","chicken"],["steak","fish","chicken"],["fish","steak","chicken"],["steak","chicken","fish"],["fish","chicken","steak"],["chicken","steak","fish"]]
print(votes)
steak = 0
chicken = 0
fish = 0

for i in range(len(votes)):
  if votes[i][0] == "steak":
    steak += 3
  elif votes[i][0] == "chicken":
    chicken += 3
  elif votes[i][0] == "fish":
    fish += 3

  if votes[i][1] == "steak":
    steak += 2
  elif votes[i][1] == "chicken":
    chicken += 2
  elif votes[i][1] == "fish":
    fish += 2

  if votes[i][2] == "steak":
    steak += 1
  elif votes[i][2] == "chicken":
    chicken += 1
  elif votes[i][2] == "fish":
    fish += 1

print(f"fish:{fish}")
print(f"steak:{steak}")
print(f"chicken:{chicken}")
```

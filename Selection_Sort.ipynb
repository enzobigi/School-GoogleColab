```python
import random

list1 = [random.randint(1, 10000) for _ in range(10000)]
print(list1)


minimumindex = 0
minimum = list1[0]


for i in (list1):
  minimum = list1[minimumindex]

  for j in range(len(list1) - minimumindex):

    if list1[j + minimumindex] < minimum:

      minimum = list1[j + minimumindex]
      q = j + minimumindex
  minimumindex += 1
  list1[minimumindex-1], list1[q] = list1[q], list1[minimumindex-1]


 # print(list1[minimumindex-1])
 # print(list1[q])
 # print(q)
 # print("\n")




print(list1)
```
